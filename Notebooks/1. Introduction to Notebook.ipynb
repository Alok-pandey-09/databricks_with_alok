## Limitation of Databricks Free Edition

In free edition you won't have access to Scala & R nor does it support legacy Resilient Distributed Datasets (RDDs). The platform is limited to a single user node, offering no GPU, no support, and no production-grade features.

### Unsupported features

The following features are not supported in Free Edition:

- R and Scala
- Custom workspace storage locations
- Online tables
- Clean rooms
- Agent Bricks
- All legacy Databricks features
- Lakebase database instances

### Administrative limitations

Free Edition accounts are meant for non-commercial use. Because of this, they have the following enterprise admin-related limits:

- One workspace and one metastore per account.
- No access to the account console or account-level APIs.
- No compliance enforcement, security customization, or private networking configurations.
- Authentication is limited to email OTP, Sign in with Google, and Sign in with Microsoft. No SSO, or SCIM support.

### Additional limitations

- Databricks may delete Free Edition accounts that are inactive for a prolonged period.
- Free Edition accounts are not covered by the Databricks support policy or Service Level Agreement (SLA).
- Free Edition accounts may not be used for commercial purposes.
- Free Edition accounts cannot become Databricks Marketplace providers.

## Magic Commands
magic commands are special instructions prefixed with `%` (for single lines) or `%%` (for entire cells) that allow you to perform tasks outside of standard Spark or PySpark code. They are primarily used for switching languages, managing files, and configuring the environment. 

### Language Magic Commands
These commands change the execution context for a specific cell to a different language (Python is the default notebook language).
- `%sql`: you can execute sql commands.
- `%md`: you can execute markdown commands.
- `%scala`: Executes Scala code.
- `%r`: Executes R code.

### System and File Management
Commands for interacting with the underlying system and Databricks Filesystem (DBFS).
- `%sh`: Runs shell (bash) commands on the driver node.
- `%fs`: Executes dbutils filesystem commands as a shorthand (e.g., `%fs ls /path).
- `%pip`: Installs Python packages directly to the notebook-scoped environment.

### Notebook Management and Documentation`:
- `%run`: Executes another notebook, allowing code modularization and the import of functions/variables.
- `%md`: Renders the cell content as Markdown for documentation purposes.

### Debugging and Utilities`:
- `%lsmagic`: Lists all available magic commands.
- `%time` : Measures the execution time of a **single line** of code.
- `%%time`: Measures the execution time of an **entire cell**.
- `%whos`: Displays information about variables in the current session.
- `%skip`: Prevents the current cell from running.

## Databricks Widgets
**Widgets = input parameters for notebooks**

````python
dbutils.widgets.text("name", "Alok")
````

````python
name = dbutils.widgets.get("name")
print(name)
````

````python
dbutils.widgets.dropdown("env", "dev", ["dev", "test", "prod"])
````

````python
env = dbutils.widgets.get("env")
print(env)
````
