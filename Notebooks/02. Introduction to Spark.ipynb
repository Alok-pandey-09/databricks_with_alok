### Lazy Evaluation

Spark optimizes data processing by identifying the most efficient physical plan to evaluate logic specified by transformations. However, Spark does not act on transformations until actions are called. Rather than evaluating each transformation in the exact order specified, Spark waits until an action triggers computation on all transformations. This is known as lazy evaluation, or lazy loading, which allows you to chain multiple operations because Spark handles their execution in a deferred manner, rather than immediately executing them when they are defined.

**Lazy evaluation means that DataFrames store logical queries as a set of instructions against a data source rather than an in-memory result. This varies drastically from eager execution, which is the model used by pandas DataFrames.**

Apache Spark uses lazy evaluation to process transformations and actions defined with DataFrames. These concepts are fundamental to understanding data processing with Spark.

Spark works in 2 phases:
1. **Transformations**→ Lazy (no execution)
2. **Actions** → Trigger execution

### Transformations

In Spark you express processing logic as transformations, which are instructions for loading and manipulating data using DataFrames. Transformation helps in building logical Plan.

Common transformations include reading data, joins, aggregations, and type casting. Eg. Select, Filter, Where, groupBy etc

#### Narrow Transformations
- After applying transformation, each partion contributes to at-most one partion.
- Partition (before and after) relationship is One-to-One 
- It does not leads to data shuffle

#### Wide Transformations
- After applying transformation, one partition contribute to ore than one partition.
- Partition (before and after) relationship is One-to-Many
- It leads to data shuffle

### Actions
- Actions instruct Spark to compute a result from a series of transformations on one or more DataFrames. It trigers the execution
Action operations return a value, and can be any of the following:

- Actions to output/view data in the console or your editor, such as `display` or `show`
- Actions to collect data (returns Row objects), such as `take(n)`, and `first` or `head`
- Actions to write to data sources, such as `saveAsTable`
- Aggregations that trigger a computation, such as `count`, `sum`

**In production data pipelines, writing data is typically the only action that should be present. All other actions interrupt query optimization and can lead to bottlenecks.**

## Structured API - Spark Dataframe
- DataFrames are the primary objects in Apache Spark. It is most common Structured API represented like a table.
- A DataFrame is a dataset organized into Rows  and named Columns. You can think of a DataFrame like a spreadsheet or a SQL table, a two-dimensional labeled data structure of a series of records (similar to rows in a table) and columns of different types.
- DataFrames has schema, which is the metadata for the columns.
- Data in DataFrames are in Partitions.
- DataFrames are **immutable**.
- DataFrames provide a rich set of functions (for example, select columns, filter, join, and aggregate) that allow you to perform common data manipulation and analysis tasks efficiently.

### DataFrame elements

- **Schema:** A schema defines the column names and types of a DataFrame. Data formats have different semantics for schema definition and enforcement. Some data sources provide schema information, while others either rely on manual schema definition or allow schema inference. Users can define schemas manually or schemas can be read from a data source.

- **Rows:** Spark represents records in a DataFrame as Row objects. While underlying data formats such as Delta Lake use columns to store data, for optimization Spark caches and shuffles data using rows.

- **Columns:** Columns in Spark are similar to columns in a spreadsheet and can represent a simple type such as a string or integer, but also complex types like array, map, or null. You can write queries that select, manipulate, or remove columns from a data source. Possible data sources include tables, views, files, or other DataFrames. Columns are never removed from a dataset or a DataFrame, they are just omitted from results through .drop transformations or omission in select statements.

### Key characteristics and features of a Spark DataFrame
- **Distributed and Scalable:** DataFrames are distributed across a cluster of machines, allowing for parallel processing of petabytes of data efficiently.
- **Schema-aware:** Unlike the lower-level RDD API, DataFrames have a defined schema (data structure, column names, and types) which allows Spark to perform optimizations. You can view a DataFrame's schema using the `.printSchema()` method.
- **Optimized Execution:** They leverage Spark's advanced optimization engines, such as the Catalyst Optimizer, which automatically generates an optimized execution plan for computations.
- **Easy-to-use API:** DataFrames provide a rich set of functions for common data analysis tasks (filtering, grouping, joining, etc.) using high-level APIs available in multiple programming languages, including Python, Scala, Java, and R.
- **SQL Integration:** DataFrames can be registered as temporary tables, allowing users to query them using standard SQL syntax.
- **Data Source Versatility:** DataFrames can be created from a wide variety of sources, such as structured data files (CSV, Parquet, JSON), Hive tables, external databases, or existing RDDs.
- **Lazy Evaluation and Immutability:** Transformations on DataFrames are lazily evaluated, meaning computation only occurs when an action (e.g., displaying results or saving output) is explicitly called. They are also immutable; operations create new DataFrames instead of modifying existing ones


### DataFrames are immutable

DataFrames are a collection of transformations and actions that are defined against one or more data sources, but ultimately Apache Spark resolves queries back to the original data sources, so the data itself is not changed, and no DataFrames are changed. In other words, DataFrames are immutable. Because of this, after performing transformations, a new DataFrame is returned that has to be saved to a variable in order to access it in subsequent operations. If you want to evaluate an intermediate step of your transformation, call an action

DataFrame is like a table, but distributed.

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DateType, ArrayType, DoubleType
import datetime
import random

schema = StructType([
    StructField("employee_id", IntegerType(), False),
    StructField("department_id", IntegerType(), False),
    StructField("gender", StringType(), True),
    StructField("birthdate", DateType(), False),
    StructField("joining_date", DateType(), False),
    StructField("skills", ArrayType(StringType()), False),
    StructField("employee_full_name", StringType(), False),
    StructField("salary", DoubleType(), False),
])

random.seed(42)
data = [
    (
        i,
        random.randint(1, 5),
        random.choice(["Male", "Female", "Other",None]),
        datetime.date(random.randint(1970, 1990), random.randint(1, 12), random.randint(1, 28)),
        datetime.date(random.randint(2010, 2026), random.randint(1, 12), random.randint(1, 28)),
        [f"Skill{j}" for j in range(1, random.randint(2, 6))],
        f"Employee_{i} Name",
        round(random.uniform(50000, 100000), 2),
    )
    for i in range(1, 31)
]

df = spark.createDataFrame(data, schema)
```

```python
from pyspark.sql.functions import col, to_date, current_date, current_timestamp, to_utc_timestamp, from_utc_timestamp, current_timezone

df.select("employee_id", "joining_date") \
    .withColumn("joining_date", to_date(col("joining_date"), 'yyyy-MM-dd')) \
    .withColumn("current_date", current_date()) \
    .withColumn("current_timestamp", current_timestamp()) \
    .withColumn("current_timestamp_UTC", to_utc_timestamp(current_timestamp(), current_timezone())) \
    .withColumn("current_timestamp_SGT", from_utc_timestamp(col("current_timestamp_UTC"), "Asia/Singapore")) \
    .show(truncate=False)
```

```python
from pyspark.sql import Row

# Create sample trades data
trades_data = [
    Row(account_id=1, trade_id=101, amount=1000),
    Row(account_id=1, trade_id=102, amount=1500),
    Row(account_id=2, trade_id=201, amount=2000),
    Row(account_id=2, trade_id=202, amount=2500),
    Row(account_id=3, trade_id=301, amount=3000)
]

trades_df = spark.createDataFrame(trades_data)
trades_df.createOrReplaceTempView("trades")

# Query 1: Grouped sum by account_id
query1 = """
SELECT account_id, SUM(amount) AS total_amount
FROM trades
GROUP BY account_id
"""
display(spark.sql(query1))

# Query 2: Window sum by account_id
query2 = """
SELECT
    account_id,
    trade_id,
    amount,
    SUM(amount) OVER (
        PARTITION BY account_id
    ) AS total_amount
FROM trades
"""
display(spark.sql(query2))
```
